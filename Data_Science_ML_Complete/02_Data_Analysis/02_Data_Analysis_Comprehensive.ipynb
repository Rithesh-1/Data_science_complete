## Why Numpy doesn't automatically convert between Integer data types

Let's create a 2-dimensional Python list 10,000 rows and 10,000 columns.

Let's also convert it into two numpy arrays, one of data type `int32` and the other with data type `int64`.

Converting to `int64` takes a bit longer because the memory allocation is higher (double). Let's also check the time required to convert between the numpy data types.

It takes so long to convert from one data type to another, because an entirely new block of memory (with twice the size) has to be allocated, and each element from the original array has to be converted and written into that block. Note that numpy cannot convert a single element, since all elements in an array must have the same data type and occupy the same number of bytes, to optimize for performance of array opertions.

Since the conversion takes a significant amount of time (and additional memory), which is why Numpy doesn't automatically perform data type conversion, since it is optimized for performance. Python, on the other hand, aims to be user-friendly, especially for integers and whole numbers, so it automatically expands or reduces the memory size as per the size of the number, and Python can do this for a single value, rather than an entire list, because there is no restriction on having different data types for different elements in a list.

That said, Numpy does perform the conversion from integer to float when either of the operands in an arithmetic operation is a float or the operation is division `/`. This behavior is programmed into Numpy because it is such a common use case.
